```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# Load dataset
data = pd.read_csv("../data/student-merge.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", data.shape)


# Columns
print("\nColumns:")
print(data.columns.tolist())


# First 5 rows
print("\nFirst 5 Rows:")
print(data.head())


# Dataset information
print("\nDataset Information:")
data.info()


# Missing values
print("\nMissing Values:")
print(data.isnull().sum())


# Duplicate rows
print("\nDuplicate Rows:", data.duplicated().sum())


# Statistical summary
print("\nStatistical Summary:")
print(data.describe())


# =====================================================
# MATHEMATICS FINAL GRADE
# =====================================================

print("\nMathematics Final Grade (G3_x):")

print("Mean:", data["G3_x"].mean())
print("Minimum:", data["G3_x"].min())
print("Maximum:", data["G3_x"].max())


# Mathematics Grade Distribution
plt.figure(figsize=(7, 4))

sns.histplot(
    data["G3_x"],
    bins=21,
    kde=True
)

plt.title("Mathematics Final Grade Distribution")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")

plt.show()


# Mathematics Study Time vs Grade
plt.figure(figsize=(7, 4))

sns.boxplot(
    data=data,
    x="studytime_x",
    y="G3_x"
)

plt.title("Mathematics Study Time vs Final Grade")
plt.xlabel("Study Time")
plt.ylabel("Final Grade")

plt.show()


# Mathematics Failures vs Grade
plt.figure(figsize=(7, 4))

sns.boxplot(
    data=data,
    x="failures_x",
    y="G3_x"
)

plt.title("Mathematics Previous Failures vs Final Grade")
plt.xlabel("Previous Failures")
plt.ylabel("Final Grade")

plt.show()


# Mathematics Absences vs Grade
plt.figure(figsize=(7, 4))

sns.scatterplot(
    data=data,
    x="absences_x",
    y="G3_x"
)

plt.title("Mathematics Absences vs Final Grade")
plt.xlabel("Absences")
plt.ylabel("Final Grade")

plt.show()


# =====================================================
# PORTUGUESE FINAL GRADE
# =====================================================

print("\nPortuguese Final Grade (G3_y):")

print("Mean:", data["G3_y"].mean())
print("Minimum:", data["G3_y"].min())
print("Maximum:", data["G3_y"].max())


# Portuguese Grade Distribution
plt.figure(figsize=(7, 4))

sns.histplot(
    data["G3_y"],
    bins=21,
    kde=True
)

plt.title("Portuguese Final Grade Distribution")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")

plt.show()


# Portuguese Study Time vs Grade
plt.figure(figsize=(7, 4))

sns.boxplot(
    data=data,
    x="studytime_y",
    y="G3_y"
)

plt.title("Portuguese Study Time vs Final Grade")
plt.xlabel("Study Time")
plt.ylabel("Final Grade")

plt.show()


# Portuguese Failures vs Grade
plt.figure(figsize=(7, 4))

sns.boxplot(
    data=data,
    x="failures_y",
    y="G3_y"
)

plt.title("Portuguese Previous Failures vs Final Grade")
plt.xlabel("Previous Failures")
plt.ylabel("Final Grade")

plt.show()


# Portuguese Absences vs Grade
plt.figure(figsize=(7, 4))

sns.scatterplot(
    data=data,
    x="absences_y",
    y="G3_y"
)

plt.title("Portuguese Absences vs Final Grade")
plt.xlabel("Absences")
plt.ylabel("Final Grade")

plt.show()


# =====================================================
# CORRELATION
# =====================================================

correlation = data.select_dtypes(
    include="number"
).corr()


# Correlation Heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()


# Mathematics correlations
print("\nCorrelation with Mathematics Final Grade (G3_x):")

print(
    correlation["G3_x"]
    .sort_values(ascending=False)
)


# Portuguese correlations
print("\nCorrelation with Portuguese Final Grade (G3_y):")

print(
    correlation["G3_y"]
    .sort_values(ascending=False)
)
```
